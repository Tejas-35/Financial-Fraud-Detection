### ***2. Importing Required Libraries***

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

pd.set_option("display.max_columns", None)
```

### **3. Load Clean Dataset**

```python
df = pd.read_csv(r"C:\Users\tejas\OneDrive\Desktop\CPMLF Capstone Projects\FRAUD DETECTION\data\processed\creditcard_cleaned.csv")
```

```python
print("Dataset shape:", df.shape)
df.head()
```

### **Sanity Checks**

```python
df.info()
```

### **4. Feature Categories**

```python
from prettytable import PrettyTable

target_col = "Class"

table = PrettyTable()
table.field_names = ["Feature Type", "Columns"]

table.add_row(["Time feature", "Time"])
table.add_row(["PCA features", "V1-V28"])
table.add_row(["Transaction", "Amount"])
table.add_row(["Target", target_col])

print(table)
```

### **5. Time Feature Engineering**

```python
df["Hour"] = (df["Time"] // 3600) % 24
```

```python
df = df.drop(columns=["Time"])
```

### **6. Transaction Amount Scaling**

```python
scaler = StandardScaler() 

df["Amount_scaled"] = scaler.fit_transform( df[["Amount"]] )
```

```python
df = df.drop(columns=["Amount"])
```

### **7. Interaction Features**

```python
df["V1_V2_interaction"] = df["V1"] * df["V2"] 
df["V3_V4_interaction"] = df["V3"] * df["V4"]
```

```python
df["Amount_per_hour"] = df["Amount_scaled"] / (df["Hour"] + 1)
```

### **8. Feature Sanity Check**

```python
df.dtypes
```

```python
print(df.shape)
```

### **9. Train test Split**

```python
X = df.drop(columns=[target_col])
y = df[target_col]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
y_train.value_counts(normalize=True)
```

### **10. Handling Class Imbalance (Training Data Only)**

```python
from imblearn.over_sampling import SMOTE
```

```python
smote = SMOTE(
    sampling_strategy=0.2,  # Resample all classes to match the majority class
    random_state=42
)

X_train_bal, y_train_bal = smote.fit_resample( 
    X_train, y_train 
    )
```

```python
y_train_bal.value_counts()
```

### **11. Final Feature Matrix Check**

```python
print("Train shape:", X_train_bal.shape)
print("Test shape:", X_test.shape)
```

```python
X_train_bal.select_dtypes(include="object").columns
```

### **12. Saving ML-Ready Dataset**

**12.1 Saving Training Dataset**

```python
train_df = pd.concat( 
    [X_train_bal, y_train_bal], 
    axis=1 
    )

train_df.to_csv("../data/processed/fraud_train_ml.csv", index=False)
```

**12.2 Saving Test Dataset**

```python
test_df = pd.concat( 
    [X_test, y_test], 
    axis=1 
    )

test_df.to_csv("../data/processed/fraud_test_ml.csv", index=False)
```

## **Feature Engineering Summary**

Key Feature Engineering Steps:
